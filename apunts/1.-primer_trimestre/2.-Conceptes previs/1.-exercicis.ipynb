3. Escriu una funció `imprimir_graf` que imprimeixi un graf.

```python
def imprimir_graf(graf):
    if isinstance(graf, list) and all(isinstance(row, list) for row in graf):
        for row in graf:
            print(row)
    elif isinstance(graf, dict):
        for vertice, vecinos in graf.items():
            print(f"{vertice}: {vecinos}")
print(imprimir_graf(graf))
```

4. Escriu una funció `grau` que donat un graf i un vèrtex retorni el grau d'aquest vèrtex.

```python
def grado(graf, vertice):
    if isinstance(graf, list) and all(isinstance(fila, list) for fila in graf):
        return sum(graf[vertice])
    elif isinstance(graf, dict):
        return len(graf.get(vertice, []))
    return -1
print({grado(graf, 2)})
```

5. Escriu una funció `veins` que donat un graf i un vèrtex retorni una llista amb els vèrtexs adjacents a aquest vèrtex.

```python
def veins(graf, vertice):
    if isinstance(graf, list) and all(isinstance(row, list) for row in graf):
        if 0 <= vertice < len(graf):
            return [i for i, es_adyacente in enumerate(graf[vertice]) if es_adyacente == 1]
        else:
            return []
    elif isinstance(graf, dict):
        return graf.get(vertice, [])
    else:
        return []

print({veins(graf, 2)})
```

8. Implementa una funció `buscar` que donat un arbre binari de búsqueda i un valor retorni el node que conté aquest valor o `None` si no el troba.

```python
def buscar(nodo, busqueda):
    if nodo is None:
        return None

    if nodo.valor == busqueda:
        return nodo

    if busqueda < nodo.valor:
        return buscar(nodo.izquierda, busqueda)
    else:
        return buscar(nodo.derecha, busqueda)
```
